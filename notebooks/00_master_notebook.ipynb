## Step 1: EDA and Feature Selection
- Loads GDHS 2022 data
- Cleans and preprocesses variables

## Step 2: Machine Learning Models
- Encodes variables (one-hot for LR, label for RF/XGBoost)

- Trains Logistic Regression, Random Forest, XGBoost, Ensemble
- Runs Stratified 5-Fold and GroupKFold cross-validation
- Checks bias-variance tradeoff
- Plots confusion matrices and ROC curves
- Computes Adjusted Odds Ratios with bootstrap CI
- Saves models to `.pkl` files

```python
%run 02_machine_learning.py
```

## Step 3: SHAP Values and Regional Analysis
